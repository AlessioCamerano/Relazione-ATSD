# Controllo Qualità del Dataset

Questo notebook esegue controlli di qualità su tutti i file di dati:
1. **Verifica valori mancanti** - Controlla la presenza di valori NULL/NaN
2. **Verifica tipi di dati** - Controlla che tutte le colonne siano numeriche

> **Nota:** I controlli vengono eseguiti su tutti i 21 file nella cartella RawData e sul file labels.txt.

---

## 1. Importazione delle librerie necessarie
Importiamo le librerie fondamentali per la manipolazione e l'analisi dei dati.

```python
import os
import pandas as pd
import numpy as np
from glob import glob

data_dir = 'RawData'
file_paths = sorted(glob(os.path.join(data_dir, 'gyro_exp*.txt')))
label_path = os.path.join(data_dir, 'labels.txt')

print(f"Trovati {len(file_paths)} file da analizzare")
```

## 2. Controllo dei valori mancanti su tutti i file
Verifichiamo la presenza di valori mancanti in tutti i file RawData e nelle label.

```python
# Analizza tutti i file gyro
print("="*60)
print("CONTROLLO VALORI MANCANTI")
print("="*60 + "\n")

all_clean = True

for file_path in file_paths:
    df = pd.read_csv(file_path, sep=' ', header=None)
    null_count = df.isnull().sum().sum()
    
    if null_count > 0:
        print(f"❌ {os.path.basename(file_path)}: {null_count} valori mancanti")
        all_clean = False
    else:
        print(f"✅ {os.path.basename(file_path)}: 0 valori mancanti")

# Controlla anche le label
df_labels = pd.read_csv(label_path, sep=' ', header=None)
label_null = df_labels.isnull().sum().sum()

if label_null > 0:
    print(f"❌ labels.txt: {label_null} valori mancanti")
    all_clean = False
else:
    print(f"✅ labels.txt: 0 valori mancanti")

print("\n" + "="*60)
if all_clean:
    print("✅ RISULTATO: Nessun valore mancante trovato in tutti i file")
else:
    print("❌ RISULTATO: Alcuni file contengono valori mancanti")
print("="*60)
```

## 3. Controllo dei tipi di dati su tutti i file
Verifichiamo che tutte le colonne siano numeriche (float64) in tutti i file.

```python
print("="*60)
print("CONTROLLO TIPI DI DATI")
print("="*60 + "\n")

all_numeric = True

for file_path in file_paths:
    df = pd.read_csv(file_path, sep=' ', header=None)
    
    # Verifica che tutte le colonne siano numeriche
    non_numeric = df.select_dtypes(exclude=['number']).columns.tolist()
    
    if len(non_numeric) > 0:
        print(f"❌ {os.path.basename(file_path)}: Colonne non numeriche trovate: {non_numeric}")
        all_numeric = False
    else:
        print(f"✅ {os.path.basename(file_path)}: Tutte le colonne sono numeriche ({df.dtypes.iloc[0]})")

# Controlla anche le label
df_labels = pd.read_csv(label_path, sep=' ', header=None)
non_numeric_labels = df_labels.select_dtypes(exclude=['number']).columns.tolist()

if len(non_numeric_labels) > 0:
    print(f"❌ labels.txt: Colonne non numeriche trovate: {non_numeric_labels}")
    all_numeric = False
else:
    print(f"✅ labels.txt: Tutte le colonne sono numeriche ({df_labels.dtypes.iloc[0]})")

print("\n" + "="*60)
if all_numeric:
    print("✅ RISULTATO: Tutti i file hanno colonne numeriche")
else:
    print("❌ RISULTATO: Alcuni file hanno colonne non numeriche")
print("="*60)
```
